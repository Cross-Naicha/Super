```python
import json
import mysql.connector
from sql_connection import sql_details
```

```python
def get_table():
    
    QUERY = "SELECT * FROM v_historic_prices;"
    SUCCESS_MESSAGE = "Tabla obtenida con exito"
    FAILURE_MESSAGE = "Error en la base de datos:"

    try:
        conn = mysql.connector.connect(**sql_details)
        cursor = conn.cursor()

        cursor.execute(QUERY)
        table = cursor.fetchall()
        print(SUCCESS_MESSAGE)

        return table

    except mysql.connector.Error as e:
        print(FAILURE_MESSAGE, e)

    finally:
        if cursor:
            cursor.close()
        if conn:
            conn.close()

def get_json(file_path="prices.json"):
    
    table = get_table()

    prices_json = dict()
    
    for row in table:
        category, product, p_price, f_price, date = row
        prices_json[product] = {'category':category,'p_price':float(p_price),'f_price':float(f_price),'date':date.strftime("%d/%m/%Y")}

    with open(file_path, "w", encoding="utf-8") as f:
        json.dump(prices_json, f, ensure_ascii=False, indent=4)
```
